# Accelerated Quantum research with AI agents and compute

## New way of doing scientific research
The small example below shows how one can combine AI agents with compute and dedicated quantum environment such as BlueQubit's platform to quickly iterate over computationally heavy experiments.
<br>To learn more about:
1. **BlueQubit Capabilities:** check out our [docs](https://app.bluequbit.io/docs) and [tutorials](https://app.bluequbit.io/tutorials)
2. **Claude Code**: best practices  [here](https://code.claude.com/docs/en/best-practices)

## Example 1: Reproducing Google's 67 qubit RCS supremacy plot for XEB

We will reproduce some small part of Google's famous RCS experiment here, by simulating lots of circuits very quickly.

1. Click "Run Example" in the top right corner to open this file in a dedicated Jupiter environment.
1. Open a new **terminal** from the Jupiter Launcher.
1. Run `claude` and log in (we assume you have a claude code account)
1. Copy and paste the following prompt into the chat window:
    <div style="background-color: #1e1e1e; color: #d4d4d4; font-family: 'Courier New', Courier, monospace; padding: 15px; border-left: 5px solid #007acc; border-radius: 0 6px 6px 0; max-width: 650px; font-size: 13px; line-height: 1.5; margin: 10px 0;">
Reproduce a mini version of Google's 67-qubit random-circuit-sampling (RCS) benchmark on BlueQubit <br>
Keep it mini: You do not need all circuits - you can use 5% of all circuits for the simulation. Try new circuit at each depth. Try using a BQ LARGE instance for the runs.<br>
Deliver. <br>
(1) Your mini plot in the style of Fig. S3a: fidelity vs cycles  <br>
(2) Google's original Fig. S3a <br>
(3) A short pdf report of the work done

    </div>
1. The last step ensures everything runs systematically:
    <div style="background-color: #e8f5e9; border-left: 5px solid #4caf50; padding: 12px; color: #1b5e20; max-width: 650px; margin-top: 10px; border-radius: 0 4px 4px 0;">
        <b>💡 Experiment Execution:</b> It will automatically queue up the runs, report live progress, generate plots, and compile a comprehensive report on everything once finished.
    </div>

The whole workflow should take ~10 minutes. You can also tell claude in the end to reproduce the whole experiment (this will require simulating ~600 circuits so might take ~40 minutes).
